## Accessing Image Paths

The table doesn't store `cached_path` as a column for portability. Instead, you can access Pixeltable's cached file location on-demand using `kamon.image.localpath`:


```python
# Access cached file path on-demand (not stored as a column)
# Note: kamon.image.localpath becomes 'image_localpath' in results (underscore, not dot)
result = kamon.select(
    kamon.description,
    kamon.image.localpath,  # Where Pixeltable cached the image
    kamon.github_url        # Original source URL
).head(3)

# Display the result
for row in result:
    print(f"Description: {row['description']}")
    print(f"Cached at: {row['image_localpath']}")
    print(f"Source URL: {row['github_url']}")
    print()
```

This approach keeps the table portable - the cached path is computed dynamically based on where Pixeltable stores files on your system.
